```python
class FilaVendedores:
    def __init__(self):
        self.fila = []

    def adicionar_vendedor(self,nome):
        """Adicionar um vendedor ao final da fila"""
        self.fila.append(nome)
        print(f"Vendedor {nome} adicionado á fila.")

    def chamar_proximo(self):
        """Chamar o proximo vendedor da fila"""
        if self.fila:
            vendedor = self.fila.pop(0)
            print(f"Próximo vendedor:{vendedor}")
            return vendedor
        else:
            print("Fila vazia!Nenhum vendedor para chamar.")
            return None

    def mostrar_fila(self):
        """Mostrar a fila atual de vendedores"""
        if self.fila:
            print("\n Fila atual:")
            for i, vendedor in enumerate(self.fila, 1):
                print(f"{i}º {vendedor}")
        else:
            print("\n Fila vazia.")

    def remover_vendedor(self,nome):
        """Remova um vendedor específico da fila"""
        if nome in self.fila:
            self.fila.remove(nome)
            print(f"{nome} removido da fila.")
        else:
            print(f"{nome} não encontrado na fila.")

    def tamanho_fila(self):
        """Retornar o número de vendedores na fila"""
        return len(self.fila)

def menu():
    """Exibir o menu principal"""
    print("\n"+"="*40)
    print("  SISTEMA DE FILA DE VENDEDORES")
    print("="*40)
    print("1. Adicionar vendedor")
    print("2. Chamar próximo vendedor")
    print("3. Mostrar fila")
    print("4. Remover vendedor")
    print("5. Ver tamanho da fila")
    print("6. Sair")
    print("="*40)

def main():
    fila = FilaVendedores()

    print(" SISTEMA DE ORGANIZAÇÃO DE FILA DE VENDEDORES")
    print("="*50)

    while True:
        menu()
        try:
            opcao = input("\nEscolha uma opção (1-6):").strip()

            if opcao == "1":
                nome = input("Nome do vendedor:").strip()
                if nome:
                    fila.adicionar_vendedor(nome)
                else:
                    print(" Nome inválido!")

            elif opcao == "2":
                fila.chamar_proximo()

            elif opcao == "3":
                fila.mostrar_fila()

            elif opcao == "4":
                if fila.tamanho_fila()>0:
                    fila.mostrar_fila()
                    nome = input("Nome do vendedor a remover:").strip()
                    if nome:
                        fila.remover_vendedor(nome)
                else:
                    print(" Fila vazia!")

            elif opcao =="5":
                tamanho = fila.tamanho_fila()
                print(f" Número de vendedores na fila:{tamanho}")

            elif opcao == "6":
                print("\n Programa encerrado!")
                break

            else:
                print(" Opção inválida! Escolha entre 1 a 6.")

        except KeyboardInterrupt:
            print("\n\n Programa interrompido pelo usuário!")
            break
        except Exception as e:
            print(f" Ocorreu um erro: {e}")


if __name__ == "__main__":
    main()
```
